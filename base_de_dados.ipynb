# Receita

```python
# vamos criar uma nova coluna
df_shows_eventos["receita"] = df_shows_eventos['Ingressos_Vendidos'] * df_shows_eventos['Preco_Medio_Ingresso']

display(df_shows_eventos[["Artista", "Cidade", "receita", "Taxa de ocupação"]])
#(ingressos vendidos / capacidade) * 100

df_shows_eventos["Taxa de ocupação"] = (df_shows_eventos["Ingressos_Vendidos"] / df_shows_eventos["Capacidade_Local"])*100
```
